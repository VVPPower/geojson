```
import pandas as pd
import numpy as np
from dataclasses import dataclass
from itertools import permutations
from pathlib import Path
from typing import List, Tuple, Optional, Dict


# -----------------------------
# CONFIG
# -----------------------------
PRODUCTS_XLSX = "Products Info.xlsx"
BOXES_XLSX    = "Carton Boxsizes.xlsx"

# Use Sheet2 for the "one order" data
PRODUCTS_SHEET_NAME = "Sheet2"
BOXES_SHEET_NAME    = None  # auto-detect

# Max number of cartons allowed for a single order (you asked 2 or 3)
MAX_BOXES_ALLOWED = 3

# Candidate filter: allow some slack above pure volume bound
# If set to 0.85, then require box_volume >= total_item_volume / 0.85 (i.e., max 85% fill)
TARGET_MAX_FILL = 0.85

# Limit candidate boxes tested (smallest volumes first)
TOP_K_CANDIDATES = 40

# If your Sheet2 contains huge quantities, you can cap expanded items to keep runtime sane.
# Set None to disable.
MAX_EXPANDED_ITEMS = 500


# -----------------------------
# Column aliasing (kept simple)
# -----------------------------
PROD_ALIASES = {
    "productidcode": "ProductIdCode",
    "productid": "ProductID",
    "itemid": "ProductID",
    "sku": "ProductID",
    "altitemid": "ProductID",

    "description": "Description",
    "itemname": "Description",
    "productname": "Description",
    "category": "Category",
    "itemgroupid": "Category",

    "length": "L", "l": "L",
    "width":  "W", "w": "W",
    "height": "H", "h": "H",

    "qty": "Qty",
    "quantity": "Qty",
    "orderqty": "Qty",
}

BOX_ALIASES = {
    "boxid": "BoxID",
    "cartonid": "BoxID",
    "carton": "BoxID",
    "sku": "BoxID",
    "style": "BoxStyle",

    "boxl": "BoxL", "boxlength": "BoxL", "length": "BoxL", "l": "BoxL",
    "boxw": "BoxW", "boxwidth":  "BoxW", "width":  "BoxW", "w": "BoxW",
    "boxh": "BoxH", "boxheight": "BoxH", "height": "H", "h": "BoxH", "depth": "BoxH",
}


def _normalize(df: pd.DataFrame, aliases: dict) -> pd.DataFrame:
    mapping = {}
    for c in df.columns:
        key = str(c).lower().strip()
        key = key.replace(" ", "").replace("-", "").replace(".", "").replace("__", "_")
        mapping[c] = aliases.get(key, c)
    return df.rename(columns=mapping)


def _clean_id_series(s: pd.Series) -> pd.Series:
    s_str = s.astype(object).astype(str).str.strip()
    nullish = s_str.str.lower().isin(["", "nan", "none", "na", "null"])
    s_str = s_str.mask(nullish, "")
    s_str = s_str.str.replace(r"\.0$", "", regex=True)
    return s_str


def load_order_items(products_path: str, sheet_name: str) -> pd.DataFrame:
    df = pd.read_excel(products_path, sheet_name=sheet_name)
    df = _normalize(df, PROD_ALIASES)

    # Require dims
    for c in ("L", "W", "H"):
        if c not in df.columns:
            raise ValueError(f"Sheet '{sheet_name}' missing required column '{c}'. Found: {list(df.columns)}")
        df[c] = pd.to_numeric(df[c], errors="coerce")
    df = df.dropna(subset=["L", "W", "H"]).copy()

    # IDs
    if "ProductIdCode" in df.columns:
        df["ProductID"] = _clean_id_series(df["ProductIdCode"])
    elif "ProductID" in df.columns:
        df["ProductID"] = _clean_id_series(df["ProductID"])
    else:
        df["ProductID"] = [f"ROW_{i}" for i in df.index]

    # Qty
    if "Qty" not in df.columns:
        df["Qty"] = 1
    df["Qty"] = pd.to_numeric(df["Qty"], errors="coerce").fillna(1).astype(int)
    df.loc[df["Qty"] <= 0, "Qty"] = 1

    # Optional text cols
    if "Category" not in df.columns:
        df["Category"] = ""
    if "Description" not in df.columns:
        df["Description"] = ""

    # Keep relevant
    return df[["ProductID", "Category", "Description", "L", "W", "H", "Qty"]].copy()


def load_boxes(boxes_path: str, sheet_name: Optional[str] = None) -> pd.DataFrame:
    xf = pd.ExcelFile(boxes_path)
    sheets = [sheet_name] if sheet_name else xf.sheet_names

    candidates = []
    for s in sheets:
        bdf = pd.read_excel(boxes_path, sheet_name=s)
        # drop Unnamed columns
        mask = ~pd.Index(bdf.columns).astype(str).str.match(r"^unnamed", case=False, na=False)
        bdf = bdf.loc[:, mask].copy()

        bdf = _normalize(bdf, BOX_ALIASES)
        if {"BoxL", "BoxW", "BoxH"}.issubset(set(bdf.columns)):
            candidates.append((s, bdf))

    if not candidates:
        raise ValueError(f"Could not find BoxL/BoxW/BoxH in {boxes_path} sheets {sheets}")

    _, boxes = candidates[0]
    for c in ("BoxL", "BoxW", "BoxH"):
        boxes[c] = pd.to_numeric(boxes[c], errors="coerce")
    boxes = boxes.dropna(subset=["BoxL", "BoxW", "BoxH"]).copy()

    if "BoxID" not in boxes.columns:
        boxes["BoxID"] = boxes.apply(lambda r: f"BOX_{r['BoxL']}x{r['BoxW']}x{r['BoxH']}", axis=1)

    boxes["BoxVol"] = boxes["BoxL"] * boxes["BoxW"] * boxes["BoxH"]
    boxes = boxes[boxes["BoxVol"] > 0].copy()

    return boxes


# -----------------------------
# Packing heuristic (3D shelf/layers)
# -----------------------------
@dataclass
class Item:
    pid: str
    l: float
    w: float
    h: float
    vol: float


def expand_items(order_df: pd.DataFrame) -> List[Item]:
    items: List[Item] = []
    for _, r in order_df.iterrows():
        pid = str(r["ProductID"])
        L, W, H = float(r["L"]), float(r["W"]), float(r["H"])
        qty = int(r["Qty"])
        for _ in range(qty):
            items.append(Item(pid=pid, l=L, w=W, h=H, vol=L * W * H))

    if MAX_EXPANDED_ITEMS is not None and len(items) > MAX_EXPANDED_ITEMS:
        # Keep the biggest items first (more important for feasibility); note this is a safety cap
        items.sort(key=lambda x: x.vol, reverse=True)
        items = items[:MAX_EXPANDED_ITEMS]

    # Sort for FFD
    items.sort(key=lambda x: x.vol, reverse=True)
    return items


def item_orientations(it: Item) -> List[Tuple[float, float, float]]:
    # all 6 rotations
    return list({(a, b, c) for (a, b, c) in permutations((it.l, it.w, it.h), 3)})


def pack_items_one_box(items: List[Item], box_dims: Tuple[float, float, float]) -> Tuple[bool, float, int]:
    """
    Returns:
      feasible (bool),
      used_volume (float),
      placed_count (int)

    Shelf/Layers heuristic:
      - Layers stacked along Z (height)
      - Rows within layer along Y (width)
      - Items placed along X (length)
    """
    BL, BW, BH = box_dims

    used_vol = 0.0
    placed = 0

    # layer state
    z = 0.0
    layer_h = 0.0

    # row state within current layer
    y = 0.0
    row_h = 0.0
    x = 0.0

    def new_layer(next_item_h: float) -> bool:
        nonlocal z, layer_h, y, row_h, x
        z += layer_h
        if z + next_item_h > BH + 1e-9:
            return False
        layer_h = 0.0
        y = 0.0
        row_h = 0.0
        x = 0.0
        return True

    def new_row(next_item_w: float) -> bool:
        nonlocal y, row_h, x
        y += row_h
        if y + next_item_w > BW + 1e-9:
            return False
        row_h = 0.0
        x = 0.0
        return True

    for it in items:
        placed_this = False

        # try all orientations (prefer larger base first to reduce fragmentation)
        oris = item_orientations(it)
        oris.sort(key=lambda t: (t[0] * t[1], -t[2]), reverse=True)

        for (l, w, h) in oris:
            # sanity: must fit within box extents
            if l > BL + 1e-9 or w > BW + 1e-9 or h > BH + 1e-9:
                continue

            # attempt place in current row
            if (x + l <= BL + 1e-9) and (y + w <= BW + 1e-9) and (z + h <= BH + 1e-9):
                # place
                x += l
                row_h = max(row_h, w)
                layer_h = max(layer_h, h)
                used_vol += (l * w * h)
                placed += 1
                placed_this = True
                break

            # try new row
            saved = (x, y, row_h, layer_h, z)
            if new_row(w):
                if (x + l <= BL + 1e-9) and (y + w <= BW + 1e-9) and (z + h <= BH + 1e-9):
                    x += l
                    row_h = max(row_h, w)
                    layer_h = max(layer_h, h)
                    used_vol += (l * w * h)
                    placed += 1
                    placed_this = True
                    break
            # revert if not placed
            x, y, row_h, layer_h, z = saved

            # try new layer
            saved = (x, y, row_h, layer_h, z)
            if new_layer(h):
                if (x + l <= BL + 1e-9) and (y + w <= BW + 1e-9) and (z + h <= BH + 1e-9):
                    x += l
                    row_h = max(row_h, w)
                    layer_h = max(layer_h, h)
                    used_vol += (l * w * h)
                    placed += 1
                    placed_this = True
                    break
            x, y, row_h, layer_h, z = saved

        if not placed_this:
            return False, used_vol, placed

    return True, used_vol, placed


def best_pack_for_box(items: List[Item], box_row: pd.Series) -> Tuple[bool, float, int, Tuple[float, float, float]]:
    """
    Try all 6 rotations of the BOX (L/W/H permutations) and return the best feasible.
    """
    dims = (float(box_row["BoxL"]), float(box_row["BoxW"]), float(box_row["BoxH"]))
    best = (False, 0.0, 0, dims)

    for bd in set(permutations(dims, 3)):
        ok, used_vol, placed = pack_items_one_box(items, bd)
        if ok and used_vol > best[1]:
            best = (ok, used_vol, placed, bd)

    return best


def boxes_needed_same_type(items: List[Item], box_dims: Tuple[float, float, float], max_boxes: int) -> Tuple[bool, int, float]:
    """
    Pack items into multiple boxes of SAME TYPE using First-Fit-Decreasing:
      - Try to pack remaining items into current box;
      - If fail, open a new box;
    Returns feasible, number_of_boxes_used, total_used_volume
    """
    remaining = items[:]
    total_used = 0.0
    used_boxes = 0

    while remaining and used_boxes < max_boxes:
        used_boxes += 1

        # Greedy: attempt to pack as many as possible into one box
        packed = []
        not_packed = []

        # We'll try to place items one-by-one; if an item can't be placed, we leave it for next box
        # This is a simplification but works well as a baseline.
        current_items = []
        for it in remaining:
            trial = current_items + [it]
            ok, used_vol, placed = pack_items_one_box(trial, box_dims)
            if ok:
                current_items.append(it)
                packed.append(it)
            else:
                not_packed.append(it)

        # If we couldn't pack ANY item in this box, fail early
        if len(packed) == 0:
            return False, used_boxes, total_used

        # Update totals and remaining
        ok, used_vol, placed = pack_items_one_box(current_items, box_dims)
        total_used += used_vol
        remaining = not_packed

    feasible = (len(remaining) == 0)
    return feasible, used_boxes, total_used


# -----------------------------
# Candidate filtering
# -----------------------------
def filter_candidate_boxes(items: List[Item], boxes: pd.DataFrame) -> pd.DataFrame:
    total_item_vol = sum(it.vol for it in items)
    longest = max(max(it.l, it.w, it.h) for it in items)

    # Lower bound by volume (allowing target fill)
    vol_lb = total_item_vol / max(TARGET_MAX_FILL, 1e-9)

    # Quick filter: box volume must exceed bound
    cand = boxes[boxes["BoxVol"] >= vol_lb].copy()

    # Additional sanity: at least one dimension of the box must be >= longest (since box can rotate)
    # (This is weak but cheap and safe.)
    cand = cand[
        (cand[["BoxL", "BoxW", "BoxH"]].max(axis=1) >= longest)
    ].copy()

    # Smallest volume first
    cand = cand.sort_values("BoxVol", ascending=True).head(TOP_K_CANDIDATES).copy()
    return cand


# -----------------------------
# MAIN
# -----------------------------
def main():
    print(f"Reading order from: {PRODUCTS_XLSX} | sheet: {PRODUCTS_SHEET_NAME}")
    order_df = load_order_items(PRODUCTS_XLSX, PRODUCTS_SHEET_NAME)

    print(f"Reading boxes from: {BOXES_XLSX}")
    boxes = load_boxes(BOXES_XLSX, BOXES_SHEET_NAME)

    items = expand_items(order_df)
    if len(items) == 0:
        raise ValueError("No items found in Sheet2 after cleaning dimensions.")

    total_item_vol = sum(it.vol for it in items)
    print(f"Items expanded: {len(items)} | Total item volume: {total_item_vol:.2f}")

    candidates = filter_candidate_boxes(items, boxes)
    if len(candidates) == 0:
        print("No candidate boxes passed the volume/dimension filter. Expanding search to all boxes...")
        candidates = boxes.sort_values("BoxVol", ascending=True).head(TOP_K_CANDIDATES).copy()

    results = []

    for _, b in candidates.iterrows():
        box_id = str(b["BoxID"])
        box_vol = float(b["BoxVol"])

        # 1) Try ONE box (best rotation)
        ok1, used1, placed1, best_dims = best_pack_for_box(items, b)
        if ok1:
            fill = used1 / box_vol
            results.append({
                "BoxID": box_id,
                "Plan": "1_box",
                "BoxesUsed": 1,
                "BoxVol": box_vol,
                "TotalBoxVol": box_vol,
                "UsedVol": used1,
                "EstimatedFill": fill,
                "BoxDimsUsed": f"{best_dims[0]}x{best_dims[1]}x{best_dims[2]}",
                "Feasible": "Y"
            })
            continue

        # 2) Try multiple boxes (same type), using best box rotation among permutations
        # We pick the best_dims from the one-box attempt as the rotation to use for multi-box packing
        # (because it’s the most promising orientation for this box type).
        feasible_multi, n_boxes, used_multi = boxes_needed_same_type(items, best_dims, MAX_BOXES_ALLOWED)

        if feasible_multi:
            total_box_vol = n_boxes * box_vol
            fill = used_multi / total_box_vol
            results.append({
                "BoxID": box_id,
                "Plan": f"{n_boxes}_boxes_same_type",
                "BoxesUsed": n_boxes,
                "BoxVol": box_vol,
                "TotalBoxVol": total_box_vol,
                "UsedVol": used_multi,
                "EstimatedFill": fill,
                "BoxDimsUsed": f"{best_dims[0]}x{best_dims[1]}x{best_dims[2]}",
                "Feasible": "Y"
            })
        else:
            results.append({
                "BoxID": box_id,
                "Plan": "no_fit_up_to_3",
                "BoxesUsed": MAX_BOXES_ALLOWED,
                "BoxVol": box_vol,
                "TotalBoxVol": MAX_BOXES_ALLOWED * box_vol,
                "UsedVol": used_multi,
                "EstimatedFill": (used_multi / (MAX_BOXES_ALLOWED * box_vol)) if box_vol > 0 else 0.0,
                "BoxDimsUsed": f"{best_dims[0]}x{best_dims[1]}x{best_dims[2]}",
                "Feasible": "N"
            })

    res_df = pd.DataFrame(results)

    # Choose BEST feasible plan:
    # Primary: fewer boxes, then smaller total volume, then higher fill
    feasible = res_df[res_df["Feasible"] == "Y"].copy()
    if len(feasible) > 0:
        feasible = feasible.sort_values(
            ["BoxesUsed", "TotalBoxVol", "EstimatedFill"],
            ascending=[True, True, False]
        )
        best = feasible.head(1).copy()
        print("\nBEST PLAN:")
        print(best.to_string(index=False))
    else:
        print("\nNo feasible plan found with <= 3 boxes among candidates.")

    out_path = Path("order_carton_selection_results.csv")
    res_df.to_csv(out_path, index=False)
    print(f"\nSaved results to: {out_path.resolve()}")


if __name__ == "__main__":
    main()
```

```
import pandas as pd
import numpy as np
from dataclasses import dataclass
from itertools import permutations
from pathlib import Path
from typing import List, Tuple, Optional, Dict


# -----------------------------
# CONFIG
# -----------------------------
PRODUCTS_XLSX = "Products Info.xlsx"
BOXES_XLSX    = "Carton Boxsizes.xlsx"

# Use Sheet2 for the "one order" data
PRODUCTS_SHEET_NAME = "Sheet2"
BOXES_SHEET_NAME    = None  # auto-detect

# Max number of cartons allowed for a single order (you asked 2 or 3)
MAX_BOXES_ALLOWED = 10

# Candidate filter: allow some slack above pure volume bound
# If set to 0.85, then require box_volume >= total_item_volume / 0.85 (i.e., max 85% fill)
TARGET_MAX_FILL = 0.85

# Limit candidate boxes tested (smallest volumes first)
TOP_K_CANDIDATES = 40

# If your Sheet2 contains huge quantities, you can cap expanded items to keep runtime sane.
# Set None to disable.
MAX_EXPANDED_ITEMS = 500


# -----------------------------
# Column aliasing (kept simple)
# -----------------------------
PROD_ALIASES = {
    "productidcode": "ProductIdCode",
    "productid": "ProductID",
    "itemid": "ProductID",
    "sku": "ProductID",
    "altitemid": "ProductID",

    "description": "Description",
    "itemname": "Description",
    "productname": "Description",
    "category": "Category",
    "itemgroupid": "Category",

    "length": "L", "l": "L",
    "width":  "W", "w": "W",
    "height": "H", "h": "H",

    "qty": "Qty",
    "quantity": "Qty",
    "orderqty": "Qty",
}

BOX_ALIASES = {
    "boxid": "BoxID",
    "cartonid": "BoxID",
    "carton": "BoxID",
    "sku": "BoxID",
    "style": "BoxStyle",

    "boxl": "BoxL", "boxlength": "BoxL", "length": "BoxL", "l": "BoxL",
    "boxw": "BoxW", "boxwidth":  "BoxW", "width":  "BoxW", "w": "BoxW",
    "boxh": "BoxH", "boxheight": "BoxH", "height": "H", "h": "BoxH", "depth": "BoxH",
}


def _normalize(df: pd.DataFrame, aliases: dict) -> pd.DataFrame:
    mapping = {}
    for c in df.columns:
        key = str(c).lower().strip()
        key = key.replace(" ", "").replace("-", "").replace(".", "").replace("__", "_")
        mapping[c] = aliases.get(key, c)
    return df.rename(columns=mapping)


def _clean_id_series(s: pd.Series) -> pd.Series:
    s_str = s.astype(object).astype(str).str.strip()
    nullish = s_str.str.lower().isin(["", "nan", "none", "na", "null"])
    s_str = s_str.mask(nullish, "")
    s_str = s_str.str.replace(r"\.0$", "", regex=True)
    return s_str


def load_order_items(products_path: str, sheet_name: str) -> pd.DataFrame:
    df = pd.read_excel(products_path, sheet_name=sheet_name)
    df = _normalize(df, PROD_ALIASES)

    # Require dims
    for c in ("L", "W", "H"):
        if c not in df.columns:
            raise ValueError(f"Sheet '{sheet_name}' missing required column '{c}'. Found: {list(df.columns)}")
        df[c] = pd.to_numeric(df[c], errors="coerce")
    df = df.dropna(subset=["L", "W", "H"]).copy()

    # IDs
    if "ProductIdCode" in df.columns:
        df["ProductID"] = _clean_id_series(df["ProductIdCode"])
    elif "ProductID" in df.columns:
        df["ProductID"] = _clean_id_series(df["ProductID"])
    else:
        df["ProductID"] = [f"ROW_{i}" for i in df.index]

    # Qty
    if "Qty" not in df.columns:
        df["Qty"] = 1
    df["Qty"] = pd.to_numeric(df["Qty"], errors="coerce").fillna(1).astype(int)
    df.loc[df["Qty"] <= 0, "Qty"] = 1

    # Optional text cols
    if "Category" not in df.columns:
        df["Category"] = ""
    if "Description" not in df.columns:
        df["Description"] = ""

    # Keep relevant
    return df[["ProductID", "Category", "Description", "L", "W", "H", "Qty"]].copy()


def load_boxes(boxes_path: str, sheet_name: Optional[str] = None) -> pd.DataFrame:
    xf = pd.ExcelFile(boxes_path)
    sheets = [sheet_name] if sheet_name else xf.sheet_names

    candidates = []
    for s in sheets:
        bdf = pd.read_excel(boxes_path, sheet_name=s)
        # drop Unnamed columns
        mask = ~pd.Index(bdf.columns).astype(str).str.match(r"^unnamed", case=False, na=False)
        bdf = bdf.loc[:, mask].copy()

        bdf = _normalize(bdf, BOX_ALIASES)
        if {"BoxL", "BoxW", "BoxH"}.issubset(set(bdf.columns)):
            candidates.append((s, bdf))

    if not candidates:
        raise ValueError(f"Could not find BoxL/BoxW/BoxH in {boxes_path} sheets {sheets}")

    _, boxes = candidates[0]
    for c in ("BoxL", "BoxW", "BoxH"):
        boxes[c] = pd.to_numeric(boxes[c], errors="coerce")
    boxes = boxes.dropna(subset=["BoxL", "BoxW", "BoxH"]).copy()

    if "BoxID" not in boxes.columns:
        boxes["BoxID"] = boxes.apply(lambda r: f"BOX_{r['BoxL']}x{r['BoxW']}x{r['BoxH']}", axis=1)

    boxes["BoxVol"] = boxes["BoxL"] * boxes["BoxW"] * boxes["BoxH"]
    boxes = boxes[boxes["BoxVol"] > 0].copy()

    return boxes


# -----------------------------
# Packing heuristic (3D shelf/layers)
# -----------------------------
@dataclass
class Item:
    pid: str
    l: float
    w: float
    h: float
    vol: float


def expand_items(order_df: pd.DataFrame) -> List[Item]:
    items: List[Item] = []
    for _, r in order_df.iterrows():
        pid = str(r["ProductID"])
        L, W, H = float(r["L"]), float(r["W"]), float(r["H"])
        qty = int(r["Qty"])
        for _ in range(qty):
            items.append(Item(pid=pid, l=L, w=W, h=H, vol=L * W * H))

    if MAX_EXPANDED_ITEMS is not None and len(items) > MAX_EXPANDED_ITEMS:
        # Keep the biggest items first (more important for feasibility); note this is a safety cap
        items.sort(key=lambda x: x.vol, reverse=True)
        items = items[:MAX_EXPANDED_ITEMS]

    # Sort for FFD
    items.sort(key=lambda x: x.vol, reverse=True)
    return items


def item_orientations(it: Item) -> List[Tuple[float, float, float]]:
    # all 6 rotations
    return list({(a, b, c) for (a, b, c) in permutations((it.l, it.w, it.h), 3)})


def pack_items_one_box(items: List[Item], box_dims: Tuple[float, float, float]) -> Tuple[bool, float, int]:
    """
    Returns:
      feasible (bool),
      used_volume (float),
      placed_count (int)

    Shelf/Layers heuristic:
      - Layers stacked along Z (height)
      - Rows within layer along Y (width)
      - Items placed along X (length)
    """
    BL, BW, BH = box_dims

    used_vol = 0.0
    placed = 0

    # layer state
    z = 0.0
    layer_h = 0.0

    # row state within current layer
    y = 0.0
    row_h = 0.0
    x = 0.0

    def new_layer(next_item_h: float) -> bool:
        nonlocal z, layer_h, y, row_h, x
        z += layer_h
        if z + next_item_h > BH + 1e-9:
            return False
        layer_h = 0.0
        y = 0.0
        row_h = 0.0
        x = 0.0
        return True

    def new_row(next_item_w: float) -> bool:
        nonlocal y, row_h, x
        y += row_h
        if y + next_item_w > BW + 1e-9:
            return False
        row_h = 0.0
        x = 0.0
        return True

    for it in items:
        placed_this = False

        # try all orientations (prefer larger base first to reduce fragmentation)
        oris = item_orientations(it)
        oris.sort(key=lambda t: (t[0] * t[1], -t[2]), reverse=True)

        for (l, w, h) in oris:
            # sanity: must fit within box extents
            if l > BL + 1e-9 or w > BW + 1e-9 or h > BH + 1e-9:
                continue

            # attempt place in current row
            if (x + l <= BL + 1e-9) and (y + w <= BW + 1e-9) and (z + h <= BH + 1e-9):
                # place
                x += l
                row_h = max(row_h, w)
                layer_h = max(layer_h, h)
                used_vol += (l * w * h)
                placed += 1
                placed_this = True
                break

            # try new row
            saved = (x, y, row_h, layer_h, z)
            if new_row(w):
                if (x + l <= BL + 1e-9) and (y + w <= BW + 1e-9) and (z + h <= BH + 1e-9):
                    x += l
                    row_h = max(row_h, w)
                    layer_h = max(layer_h, h)
                    used_vol += (l * w * h)
                    placed += 1
                    placed_this = True
                    break
            # revert if not placed
            x, y, row_h, layer_h, z = saved

            # try new layer
            saved = (x, y, row_h, layer_h, z)
            if new_layer(h):
                if (x + l <= BL + 1e-9) and (y + w <= BW + 1e-9) and (z + h <= BH + 1e-9):
                    x += l
                    row_h = max(row_h, w)
                    layer_h = max(layer_h, h)
                    used_vol += (l * w * h)
                    placed += 1
                    placed_this = True
                    break
            x, y, row_h, layer_h, z = saved

        if not placed_this:
            return False, used_vol, placed

    return True, used_vol, placed


def best_pack_for_box(items: List[Item], box_row: pd.Series) -> Tuple[bool, float, int, Tuple[float, float, float]]:
    """
    Try all 6 rotations of the BOX (L/W/H permutations) and return the best feasible.
    """
    dims = (float(box_row["BoxL"]), float(box_row["BoxW"]), float(box_row["BoxH"]))
    best = (False, 0.0, 0, dims)

    for bd in set(permutations(dims, 3)):
        ok, used_vol, placed = pack_items_one_box(items, bd)
        if ok and used_vol > best[1]:
            best = (ok, used_vol, placed, bd)

    return best


def boxes_needed_same_type(items: List[Item], box_dims: Tuple[float, float, float], max_boxes: int) -> Tuple[bool, int, float]:
    """
    Pack items into multiple boxes of SAME TYPE using First-Fit-Decreasing:
      - Try to pack remaining items into current box;
      - If fail, open a new box;
    Returns feasible, number_of_boxes_used, total_used_volume
    """
    remaining = items[:]
    total_used = 0.0
    used_boxes = 0

    while remaining and used_boxes < max_boxes:
        used_boxes += 1

        # Greedy: attempt to pack as many as possible into one box
        packed = []
        not_packed = []

        # We'll try to place items one-by-one; if an item can't be placed, we leave it for next box
        # This is a simplification but works well as a baseline.
        current_items = []
        for it in remaining:
            trial = current_items + [it]
            ok, used_vol, placed = pack_items_one_box(trial, box_dims)
            if ok:
                current_items.append(it)
                packed.append(it)
            else:
                not_packed.append(it)

        # If we couldn't pack ANY item in this box, fail early
        if len(packed) == 0:
            return False, used_boxes, total_used

        # Update totals and remaining
        ok, used_vol, placed = pack_items_one_box(current_items, box_dims)
        total_used += used_vol
        remaining = not_packed

    feasible = (len(remaining) == 0)
    return feasible, used_boxes, total_used


# -----------------------------
# Candidate filtering
# -----------------------------
def filter_candidate_boxes(items: List[Item], boxes: pd.DataFrame) -> pd.DataFrame:
    total_item_vol = sum(it.vol for it in items)
    longest = max(max(it.l, it.w, it.h) for it in items)

    # Lower bound by volume (allowing target fill)
    vol_lb = total_item_vol / max(TARGET_MAX_FILL, 1e-9)

    # Quick filter: box volume must exceed bound
    cand = boxes[boxes["BoxVol"] >= vol_lb].copy()

    # Additional sanity: at least one dimension of the box must be >= longest (since box can rotate)
    # (This is weak but cheap and safe.)
    cand = cand[
        (cand[["BoxL", "BoxW", "BoxH"]].max(axis=1) >= longest)
    ].copy()

    # Smallest volume first
    cand = cand.sort_values("BoxVol", ascending=True).head(TOP_K_CANDIDATES).copy()
    return cand


# -----------------------------
# MAIN
# -----------------------------
def main():
    print(f"Reading order from: {PRODUCTS_XLSX} | sheet: {PRODUCTS_SHEET_NAME}")
    order_df = load_order_items(PRODUCTS_XLSX, PRODUCTS_SHEET_NAME)

    print(f"Reading boxes from: {BOXES_XLSX}")
    boxes = load_boxes(BOXES_XLSX, BOXES_SHEET_NAME)

    items = expand_items(order_df)
    if len(items) == 0:
        raise ValueError("No items found in Sheet2 after cleaning dimensions.")

    total_item_vol = sum(it.vol for it in items)
    print(f"Items expanded: {len(items)} | Total item volume: {total_item_vol:.2f}")

    candidates = filter_candidate_boxes(items, boxes)
    if len(candidates) == 0:
        print("No candidate boxes passed the volume/dimension filter. Expanding search to all boxes...")
        candidates = boxes.sort_values("BoxVol", ascending=True).head(TOP_K_CANDIDATES).copy()

    results = []

    for _, b in candidates.iterrows():
        box_id = str(b["BoxID"])
        box_vol = float(b["BoxVol"])

        # 1) Try ONE box (best rotation)
        ok1, used1, placed1, best_dims = best_pack_for_box(items, b)
        if ok1:
            fill = used1 / box_vol
            results.append({
                "BoxID": box_id,
                "Plan": "1_box",
                "BoxesUsed": 1,
                "BoxVol": box_vol,
                "TotalBoxVol": box_vol,
                "UsedVol": used1,
                "EstimatedFill": fill,
                "BoxDimsUsed": f"{best_dims[0]}x{best_dims[1]}x{best_dims[2]}",
                "Feasible": "Y"
            })
            continue

        # 2) Try multiple boxes (same type), using best box rotation among permutations
        # We pick the best_dims from the one-box attempt as the rotation to use for multi-box packing
        # (because it’s the most promising orientation for this box type).
        feasible_multi, n_boxes, used_multi = boxes_needed_same_type(items, best_dims, MAX_BOXES_ALLOWED)

        if feasible_multi:
            total_box_vol = n_boxes * box_vol
            fill = used_multi / total_box_vol
            results.append({
                "BoxID": box_id,
                "Plan": f"{n_boxes}_boxes_same_type",
                "BoxesUsed": n_boxes,
                "BoxVol": box_vol,
                "TotalBoxVol": total_box_vol,
                "UsedVol": used_multi,
                "EstimatedFill": fill,
                "BoxDimsUsed": f"{best_dims[0]}x{best_dims[1]}x{best_dims[2]}",
                "Feasible": "Y"
            })
        else:
            results.append({
                "BoxID": box_id,
                "Plan": "no_fit_up_to_" + str(MAX_BOXES_ALLOWED),
                "BoxesUsed": MAX_BOXES_ALLOWED,
                "BoxVol": box_vol,
                "TotalBoxVol": MAX_BOXES_ALLOWED * box_vol,
                "UsedVol": used_multi,
                "EstimatedFill": (used_multi / (MAX_BOXES_ALLOWED * box_vol)) if box_vol > 0 else 0.0,
                "BoxDimsUsed": f"{best_dims[0]}x{best_dims[1]}x{best_dims[2]}",
                "Feasible": "N"
            })

    res_df = pd.DataFrame(results)

    # Choose BEST feasible plan:
    # Primary: fewer boxes, then smaller total volume, then higher fill
    feasible = res_df[res_df["Feasible"] == "Y"].copy()
    if len(feasible) > 0:
        feasible = feasible.sort_values(
            ["BoxesUsed", "TotalBoxVol", "EstimatedFill"],
            ascending=[True, True, False]
        )
        best = feasible.head(1).copy()
        print("\nBEST PLAN:")
        print(best.to_string(index=False))
    else:
        print(f"\nNo feasible plan found with <= {MAX_BOXES_ALLOWED} boxes among candidates.")

    out_path = Path("order_carton_selection_results.csv")
    res_df.to_csv(out_path, index=False)
    print(f"\nSaved results to: {out_path.resolve()}")


if __name__ == "__main__":
    main()
```
